```python
import pandas as pd
import numpy as np

# Show all columns when printing
pd.set_option("display.width", 250)
pd.set_option("display.max_columns", 50)

# ---------------------------------------------------------------
# TASK 1: Load data, first 10 / last 10 rows, rows and columns
# ---------------------------------------------------------------
orders = pd.read_csv("Pakistan_Largest_Ecommerce_Dataset.csv")

print("First 10 records:")
print(orders.head(10))

print("Last 10 records:")
print(orders.tail(10))   # these are empty because the file has blank rows at the end

print("Total rows:", orders.shape[0])
print("Total columns:", orders.shape[1])

# ---------------------------------------------------------------
# TASK 2: Column names and data types
# ---------------------------------------------------------------
print(orders.dtypes)

numerical = ["price", "qty_ordered", "grand_total", "discount_amount", "item_id", "Customer ID", "Year", "Month"]
categorical = ["status", "sku", "category_name_1", "sales_commission_code", "payment_method", "BI Status", "FY", "increment_id"]
date_related = ["created_at", "Working Date", "Customer Since", "M-Y", "Year", "Month"]

print("Numerical columns:", numerical)
print("Categorical/Text columns:", categorical)
print("Date-related columns:", date_related)

# ---------------------------------------------------------------
# TASK 3: Separate Series (first 10 values)
# ---------------------------------------------------------------
price = orders["price"]
quantity = orders["qty_ordered"]
payment = orders["payment_method"]
category = orders["category_name_1"]

print(price.head(10))
print(quantity.head(10))
print(payment.head(10))
print(category.head(10))

# ---------------------------------------------------------------
# TASK 4: Basic statistics
# ---------------------------------------------------------------
print("Total orders (rows):", orders["item_id"].count())
print("Total quantity ordered:", orders["qty_ordered"].sum())
print("Average product price:", orders["price"].mean())
print("Maximum product price:", orders["price"].max())
print("Minimum product price:", orders["price"].min())

# ---------------------------------------------------------------
# TASK 5: Missing values (only columns that have missing values)
# ---------------------------------------------------------------
missing = orders.isnull().sum()
print(missing[missing > 0])

# ---------------------------------------------------------------
# TASK 6: Missing percentage table
# ---------------------------------------------------------------
missing_df = pd.DataFrame()
missing_df["Column Name"] = orders.columns
missing_df["Missing Count"] = orders.isnull().sum().values
missing_df["Missing Percentage"] = (orders.isnull().sum().values / len(orders)) * 100
print(missing_df)

# ---------------------------------------------------------------
# TASK 7: Handle missing values
# ---------------------------------------------------------------
cleaned_df = orders.copy()

# 1) Remove rows that are completely empty (blank rows at the end of the file)
cleaned_df = cleaned_df.dropna(how="all")

# 2) Remove columns that are completely empty (Unnamed: 21 to 25)
cleaned_df = cleaned_df.dropna(axis=1, how="all")

# 3) The file uses \N to mean "no value". Change it to a real missing value.
cleaned_df = cleaned_df.replace("\\N", np.nan)

# 4) Remove extra spaces in column names (" MV " becomes "MV")
cleaned_df.columns = cleaned_df.columns.str.strip()

# 5) Text columns: fill with "Unknown" (we cannot guess a category)
cleaned_df["category_name_1"] = cleaned_df["category_name_1"].fillna("Unknown")
cleaned_df["status"] = cleaned_df["status"].fillna("Unknown")
cleaned_df["sku"] = cleaned_df["sku"].fillna("Unknown")

# 6) Commission code: empty means no commission was given
cleaned_df["sales_commission_code"] = cleaned_df["sales_commission_code"].fillna("No Commission")

# 7) Customer ID is needed for merging, so drop the few rows without it
cleaned_df = cleaned_df.dropna(subset=["Customer ID"])
cleaned_df["Customer ID"] = cleaned_df["Customer ID"].astype(int)

print("Missing values left:", cleaned_df.isnull().sum().sum())

# ---------------------------------------------------------------
# TASK 8: original_df vs cleaned_df
# ---------------------------------------------------------------
original_df = orders.copy()

print("Rows before cleaning:", len(original_df))
print("Rows after cleaning:", len(cleaned_df))
print("Missing values before cleaning:", original_df.isnull().sum().sum())
print("Missing values after cleaning:", cleaned_df.isnull().sum().sum())

# ---------------------------------------------------------------
# TASK 9: Split into two parts and join with pd.concat()
# ---------------------------------------------------------------
half = len(cleaned_df) // 2
orders_part1 = cleaned_df.iloc[:half]
orders_part2 = cleaned_df.iloc[half:]

combined = pd.concat([orders_part1, orders_part2])

print("Part 1 rows:", len(orders_part1))
print("Part 2 rows:", len(orders_part2))
print("After concat:", len(combined))
print("Same as cleaned_df?", len(combined) == len(cleaned_df))

# ---------------------------------------------------------------
# TASK 10: Two category DataFrames and combined_products
# ---------------------------------------------------------------
electronics_df = cleaned_df[cleaned_df["category_name_1"] == "Mobiles & Tablets"]
fashion_df = cleaned_df[cleaned_df["category_name_1"] == "Men's Fashion"]

combined_products = pd.concat([electronics_df, fashion_df])

print("electronics_df:", len(electronics_df))
print("fashion_df:", len(fashion_df))
print("combined_products:", len(combined_products))

# ---------------------------------------------------------------
# TASK 11: pd.concat(axis=1)
# ---------------------------------------------------------------
df_a = cleaned_df[["item_id", "price", "qty_ordered"]].head(100)
df_b = cleaned_df[["category_name_1", "payment_method"]].head(100)

side_by_side = pd.concat([df_a, df_b], axis=1)
print(side_by_side.head())
print("Shape:", side_by_side.shape)

# ---------------------------------------------------------------
# TASK 12: customers and orders DataFrames, find the common key
# ---------------------------------------------------------------
customers = cleaned_df[["Customer ID", "Customer Since"]].drop_duplicates(subset="Customer ID")
orders = cleaned_df[["item_id", "Customer ID", "category_name_1", "payment_method", "price", "qty_ordered", "status"]]

print("customers:", customers.shape)
print("orders:", orders.shape)
print("Common column (key): Customer ID")

# ---------------------------------------------------------------
# TASK 13: INNER merge
# ---------------------------------------------------------------
inner = pd.merge(customers, orders, on="Customer ID", how="inner")
print("Rows after INNER merge:", len(inner))

# ---------------------------------------------------------------
# TASK 14: LEFT merge and compare
# ---------------------------------------------------------------
left = pd.merge(customers, orders, on="Customer ID", how="left")
print("INNER rows:", len(inner))
print("LEFT rows:", len(left))


# ---------------------------------------------------------------
# TASK 15: inner_merged and left_merged (with unmatched records)
# ---------------------------------------------------------------
# To see a real difference, make the two tables only partly overlap:
customers_part = customers[customers["Customer ID"] <= 90000]   # customers 1 to 90000
orders_part = orders[orders["Customer ID"] > 10000]             # orders of customers above 10000

inner_merged = pd.merge(customers_part, orders_part, on="Customer ID", how="inner")
left_merged = pd.merge(customers_part, orders_part, on="Customer ID", how="left")

print("inner_merged rows:", len(inner_merged))
print("left_merged rows:", len(left_merged))

# Customers that have no order (customer ID 1 to 10000)
no_order = customers_part[~customers_part["Customer ID"].isin(orders_part["Customer ID"])]
print("Customers with no matching order:", len(no_order))

# Orders that have no customer record (customer ID above 90000)
no_customer = orders_part[~orders_part["Customer ID"].isin(customers_part["Customer ID"])]
print("Orders with no matching customer:", len(no_customer))

# ---------------------------------------------------------------
# TASK 16: .join() using the index
# ---------------------------------------------------------------
customer_data = customers.set_index("Customer ID")
order_data = orders.set_index("Customer ID")

joined = order_data.join(customer_data)
print(joined.head())
print("Rows after join:", len(joined))

# ---------------------------------------------------------------
# TASK 17: Same ID but different column names
# ---------------------------------------------------------------
customers_new = customers.rename(columns={"Customer ID": "cust_id"})

# Method 1: use left_on and right_on
result1 = pd.merge(orders, customers_new, left_on="Customer ID", right_on="cust_id")
print(result1.head())

# Method 2: rename the column so both have the same name, then use on=
customers_same = customers_new.rename(columns={"cust_id": "Customer ID"})
result2 = pd.merge(orders, customers_same, on="Customer ID")
print("Method 1 rows:", len(result1), "| Method 2 rows:", len(result2))

# ---------------------------------------------------------------
# TASK 18: Payment methods
# ---------------------------------------------------------------
print("Unique payment methods:", cleaned_df["payment_method"].unique())

payment_counts = cleaned_df["payment_method"].value_counts()
print(payment_counts)

print("Most used payment method:", payment_counts.idxmax(), "->", payment_counts.max(), "orders")

# ---------------------------------------------------------------
# TASK 19: Product categories
# ---------------------------------------------------------------
print("Unique categories:", cleaned_df["category_name_1"].unique())

print("Orders per category:")
print(cleaned_df["category_name_1"].value_counts())

print("Total quantity per category:")
print(cleaned_df.groupby("category_name_1")["qty_ordered"].sum())

print("Average price per category:")
print(cleaned_df.groupby("category_name_1")["price"].mean())

# ---------------------------------------------------------------
# TASK 20: business_report
# ---------------------------------------------------------------
business_report = cleaned_df.groupby("category_name_1").agg(
    Orders=("item_id", "count"),
    Quantity_Sold=("qty_ordered", "sum"),
    Average_Price=("price", "mean"),
)
business_report = business_report.reset_index()
business_report.columns = ["Category", "Orders", "Quantity Sold", "Average Price"]
business_report["Average Price"] = business_report["Average Price"].round(2)
business_report = business_report.sort_values("Orders", ascending=False)

print(business_report)
```
